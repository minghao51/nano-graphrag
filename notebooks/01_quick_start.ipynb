# Quick Start - Get running in 5 minutes

Insert a document into a knowledge graph, query it, and inspect the
resulting graph - with just a few lines of code.

**Prerequisites:** An OpenRouter API key (or any LiteLLM-supported provider)
exported via `dotenvx` - see **Configuration** notebook for alternatives.

**Run:**
```bash
dotenvx run -- uv run jupyter nbconvert --to notebook --execute --inplace notebooks/01_quick_start.ipynb
```


````python
import logging
from pathlib import Path

logging.basicConfig(level=logging.WARNING)
logging.getLogger("nano-graphrag").setLevel(logging.WARNING)

WORKING_DIR = Path("./_cache/01_quick_start")
WORKING_DIR.mkdir(parents=True, exist_ok=True)
````

## Load sample text

````python
with open("../tests/fixtures/mock_data.txt", encoding="utf-8-sig") as f:
    lines = f.readlines()
text = "".join(lines[:400])
print(f"Loaded {len(text):,} chars ({len(text.splitlines())} lines)")
````

## Initialize GraphRAG

````python
from nano_graphrag import GraphRAG, QueryParam

rag = GraphRAG(
    working_dir=str(WORKING_DIR),
    enable_llm_cache=True,
)
print(f"LLM model:      {rag.llm_model}")
print(f"Embedding model: {rag.embedding_model}")
````

## Insert - chunk, extract entities, build graph, detect communities

````python
import time
print("Indexing...")
start = time.time()
await rag.ainsert(text)
elapsed = time.time() - start
print(f"Done in {elapsed:.1f}s")
````

## Query - global mode (aggregates community reports)

````python
result = await rag.aquery(
    "What are the main themes in A Christmas Carol?",
    param=QueryParam(mode="global"),
)
print(result)
````

## Query - local mode (entity-level retrieval)

````python
result = await rag.aquery(
    "What is the relationship between Scrooge and Bob Cratchit?",
    param=QueryParam(mode="local"),
)
print(result)
````

## Graph overview (textual)

````python
import networkx as nx
from collections import Counter

graphml_path = WORKING_DIR / "graph_chunk_entity_relation.graphml"
G = nx.read_graphml(str(graphml_path))

node_types = Counter()
for n, data in G.nodes(data=True):
    node_types[data.get("entity_type", "UNKNOWN")] += 1

print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")
print(f"Entity types: {dict(node_types.most_common())}")
print(f"Top 5 entities by degree:")
for nid, deg in sorted(G.degree(), key=lambda x: x[1], reverse=True)[:5]:
    name = G.nodes[nid].get("entity_name", nid)
    etype = G.nodes[nid].get("entity_type", "?")
    print(f"  {name:30s} ({etype:15s}) degree={deg}")
````
