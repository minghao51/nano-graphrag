# Configuration - Three ways to configure GraphRAG

Learn kwargs, YAML, environment variables, and a practical Ollama example.

**Run:**
```bash
dotenvx run -- uv run jupyter nbconvert --to notebook --execute --inplace notebooks/03_configuration.ipynb
```


````python
import logging
from pathlib import Path

logging.basicConfig(level=logging.WARNING)
logging.getLogger("nano-graphrag").setLevel(logging.WARNING)

WORKING_DIR = Path("./_cache/03_configuration")
WORKING_DIR.mkdir(parents=True, exist_ok=True)
````

## 1. Inline kwargs (simple, self-contained)

````python
from nano_graphrag import GraphRAG
from nano_graphrag.base import GraphRAGConfig

config_yaml = str(WORKING_DIR / "custom_settings.yaml")
Path(config_yaml).write_text(
    f"""
working_dir: '{WORKING_DIR / "yaml"}'
llm_model: openai/gpt-4o-mini
enable_llm_cache: true
"""
)
config = GraphRAGConfig.from_yaml(config_yaml)
rag2 = GraphRAG(**config.to_dict())
print(f"rag2 created from YAML: {config_yaml}")

# Or from_env shorthand:
#   config = GraphRAGConfig.from_env()
#   rag = GraphRAG(config)
````

## 3. Environment variables (for CI/deployment)

````python
# GraphRAG reads API keys from the environment (injected via dotenvx).
# For config fields, use inline kwargs or YAML as shown above.
#
# Your API keys (OPENAI_API_KEY, OPENROUTER_API_KEY, etc.) are
# injected by dotenvx at runtime. Set them in .env or export them directly.

rag3 = GraphRAG(
    working_dir=str(WORKING_DIR / "env"),
    enable_llm_cache=True,
)
print("rag3 reads API keys from env vars (injected by dotenvx)")
````

## 4. Practical example - local model via Ollama

````python
# To use a local model served by Ollama, set these env vars:
#
#   LLM_MODEL=ollama/llama3.2
#   EMBEDDING_MODEL=ollama/nomic-embed-text
#   LLM_BASE_URL=http://localhost:11434
#
# Then run: dotenvx run -- uv run python your_script.py
#
# The equivalent inline config:
rag_ollama = GraphRAG(
    working_dir=str(WORKING_DIR / "ollama"),
    llm_model="ollama/llama3.2",
    embedding_model="ollama/nomic-embed-text",
    llm_api_base="http://localhost:11434",
    enable_llm_cache=True,
)
print("rag_ollama configured for local Ollama")
````

## Config hierarchy (lower number = higher priority)

````python

print("Config resolution order (1 = highest priority):")
print("  1. Inline kwargs          - passed directly to GraphRAG(...)")
print("  2. YAML file              - loaded via GraphRAG.from_yaml()")
print("  3. Environment variables  - read from os.environ on init")
print("  4. Defaults               - from config/settings.yaml")
print()
print("chunk_size=2000 inline overrides any YAML or env value.")
````
